# First 5 Rows of Each ESCO CSV

This notebook displays the first 5 rows from every CSV file in `backend_matching/esco`.


```python
from pathlib import Path
import pandas as pd
from IPython.display import display, Markdown

# Robust path detection from common notebook launch locations
CANDIDATE_DIRS = [
    Path("backend_matching/data/esco"),
    Path("backend_matching/esco"),
    Path("data/esco"),
    Path("esco"),
    Path("../data/esco"),
    Path("../esco"),
    Path("../backend_matching/data/esco"),
    Path("../backend_matching/esco"),
    Path("../../backend_matching/data/esco"),
    Path("../../backend_matching/esco"),
]

# Also search upward from current working directory for potential ESCO folders
cwd = Path.cwd()
for parent in [cwd] + list(cwd.parents):
    CANDIDATE_DIRS.append(parent / "backend_matching" / "data" / "esco")
    CANDIDATE_DIRS.append(parent / "backend_matching" / "esco")

# Keep order but remove duplicates
seen = set()
unique_candidates = []
for p in CANDIDATE_DIRS:
    key = str(p)
    if key not in seen:
        unique_candidates.append(p)
        seen.add(key)

BASE_DIR = next((p for p in unique_candidates if p.exists() and p.is_dir()), None)
if BASE_DIR is None:
    raise FileNotFoundError(
        "Could not find the ESCO CSV directory. Checked: " + ", ".join(str(p) for p in unique_candidates)
    )

CSV_FILES = sorted(BASE_DIR.glob("*.csv"))

print(f"Using CSV directory: {BASE_DIR.resolve()}")
print(f"Found CSV files: {len(CSV_FILES)}")
```

```python
for csv_file in CSV_FILES:
    display(Markdown(f"## `{csv_file.name}`"))
    df = pd.read_csv(csv_file, nrows=5)
    display(df)
```

## ILOSTAT and ISCO Datasets (First 10 Rows)

This section shows the first 10 rows from the ILOSTAT and ISCO datasets in tabular form.

```python
# Display first 10 rows from ILOSTAT and ISCO datasets
from pathlib import Path
import pandas as pd
from IPython.display import display, Markdown

candidate_roots = [Path("."), Path(".."), Path("../.."), Path("../../..")]

ilostat_file = None
isco_file = None

for root in candidate_roots:
    ilostat_candidate = root / "backend_matching/data/ilostat/EAR_EMTA_SEX_OCU_NB_A-filtered-2026-04-26.csv"
    isco_candidate = root / "backend_matching/data/isco/ISCO-08 EN Structure and definitions.csv"

    if ilostat_file is None and ilostat_candidate.exists():
        ilostat_file = ilostat_candidate
    if isco_file is None and isco_candidate.exists():
        isco_file = isco_candidate

if ilostat_file is None:
    raise FileNotFoundError("Could not find ILOSTAT CSV file.")
if isco_file is None:
    raise FileNotFoundError("Could not find ISCO CSV file.")

display(Markdown("## `EAR_EMTA_SEX_OCU_NB_A-filtered-2026-04-26.csv` (ILOSTAT)"))
ilostat_df = pd.read_csv(ilostat_file, nrows=10)
display(ilostat_df)

display(Markdown("## `ISCO-08 EN Structure and definitions.csv` (ISCO)"))
isco_df = pd.read_csv(isco_file, nrows=10)
display(isco_df)
```
